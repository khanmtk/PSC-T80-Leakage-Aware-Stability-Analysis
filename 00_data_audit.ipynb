```python
# ==========================================
# 00_data_audit.ipynb
# ==========================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', 200)

# Current project directory
WORK_DIR = r"C:\Users\khanm\OneDrive\Desktop\3rd model 18-03-2026"

# Output folder for this notebook
ARTIFACT_DIR = os.path.join(WORK_DIR, "artifacts", "00_audit")

os.makedirs(ARTIFACT_DIR, exist_ok=True)

print("Working directory:", WORK_DIR)
print("Directory exists:", os.path.exists(WORK_DIR))
print("Artifact directory:", ARTIFACT_DIR)
```

```python
# ==========================================
# LOAD RAW PEROVSKITE DATABASE
# ==========================================

file_path = r"C:\Users\khanm\OneDrive\Desktop\3rd model 18-03-2026\Perovskite_database_content_all_data.csv"

df = pd.read_csv(file_path)

print("File loaded successfully")
print("Shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())
```

```python
print("Total columns:", len(df.columns))

col_df = pd.DataFrame({
    "column": df.columns,
    "dtype": df.dtypes.values
})

col_df.to_csv(os.path.join(ARTIFACT_DIR, "column_list.csv"), index=False)

col_df.head(20)
```

```python
t80_cols = [c for c in df.columns if "T80" in c or "t80" in c]

print("T80-related columns:")
print(t80_cols)

pd.DataFrame({"T80_columns": t80_cols}).to_csv(
    os.path.join(ARTIFACT_DIR, "t80_columns.csv"), index=False
)
```

```python
if len(t80_cols) == 0:
    print("❌ No T80 column found")
else:
    for col in t80_cols:
        valid = df[col].notna().sum()
        print(f"{col}: {valid} non-null values")
```

```python
doi_cols = [c for c in df.columns if "DOI" in c]

print("DOI columns:", doi_cols)

if len(doi_cols) > 0:
    doi_col = doi_cols[0]
    
    n_groups = df[doi_col].nunique()
    counts = df.groupby(doi_col).size()
    
    print("Number of DOI groups:", n_groups)
    print("Avg samples per DOI:", counts.mean())
    print("Median samples per DOI:", counts.median())
    
    counts.describe()
    
    counts.to_csv(os.path.join(ARTIFACT_DIR, "doi_group_sizes.csv"))
```

```python
arch_cols = [c for c in df.columns if "architecture" in c.lower()]

print("Architecture columns:", arch_cols)

if len(arch_cols) > 0:
    arch_col = arch_cols[0]
    
    dist = df[arch_col].value_counts(dropna=False)
    print(dist)
    
    dist.to_csv(os.path.join(ARTIFACT_DIR, "architecture_distribution.csv"))
```

```python
missing = df.isnull().mean().sort_values(ascending=False)

missing_df = pd.DataFrame({
    "column": missing.index,
    "missing_fraction": missing.values
})

missing_df.to_csv(os.path.join(ARTIFACT_DIR, "missingness.csv"), index=False)

missing_df.head(20)
```

```python
dup_count = df.duplicated().sum()
print("Duplicate rows:", dup_count)
```

```python
num_df = df.select_dtypes(include=np.number)

summary = num_df.describe().T

summary.to_csv(os.path.join(ARTIFACT_DIR, "numeric_summary.csv"))

summary.head(20)
```

```python
leak_keywords = [
    "Ref_", "DOI", "author", "journal",
    "filename", "comment", "ID", "name"
]

leak_cols = [c for c in df.columns if any(k.lower() in c.lower() for k in leak_keywords)]

print("Potential metadata / leakage columns:")
print(leak_cols)

pd.DataFrame({"leakage_candidates": leak_cols}).to_csv(
    os.path.join(ARTIFACT_DIR, "leakage_candidates.csv"), index=False
)
```

```python
# Pick first T80 column automatically
if len(t80_cols) > 0:
    target_col = t80_cols[0]
    
    plt.figure(figsize=(6,4))
    df[target_col].dropna().hist(bins=50)
    plt.title(f"Distribution of {target_col}")
    plt.xlabel("T80")
    plt.ylabel("Count")
    plt.grid(True)
    
    plt.savefig(os.path.join(ARTIFACT_DIR, "t80_distribution.png"),dpi=600)
    plt.show()
```

```python
print("========== FINAL AUDIT SUMMARY ==========")

print("Total samples:", len(df))
print("Total features:", len(df.columns))

if len(t80_cols) > 0:
    print("Target column:", t80_cols[0])

if len(doi_cols) > 0:
    print("DOI column:", doi_cols[0])
    print("DOI groups:", df[doi_cols[0]].nunique())

print("Artifacts saved in:", ARTIFACT_DIR)
```

```python
if len(t80_cols) > 0:
    target_col = t80_cols[0]

    data = df[target_col].dropna()
    data = data[data > 0]

  
    upper_limit = data.quantile(0.95)

    plt.figure(figsize=(6,4))

    plt.hist(data, bins=50, edgecolor='black', alpha=0.8)

    plt.xlim(0, upper_limit)

    plt.title("Lifetime Zoomed to 95th Percentile")
    plt.xlabel("T80")
    plt.ylabel("Count")

    plt.grid(True, linestyle="--", linewidth=0.5, alpha=0.6)
    plt.legend()
    plt.tight_layout()

    plt.savefig(os.path.join(ARTIFACT_DIR, "t80_log10_distribution.png"), dpi=600)
    plt.show()
```

```python
if len(t80_cols) > 0:
    target_col = t80_cols[0]

    data = df[target_col].dropna()
    data = data[data > 0]

    lower_limit = data.quantile(0.95)

    plt.figure(figsize=(6,4))

    high_data = data[data >= lower_limit]

    plt.hist(high_data, bins=50, edgecolor='black', alpha=0.8)

    plt.title(f"High T80 Values (Top 5%)")
    plt.xlabel("T80")
    plt.ylabel("Count")

    plt.grid(True, linestyle="--", linewidth=0.5, alpha=0.6)
    plt.tight_layout()

    plt.savefig(os.path.join(ARTIFACT_DIR, "t80_high_values.png"), dpi=600)
    plt.show()
```

```python
# ============================================================
# FIGURE 1(c) — DISTRIBUTION OF ln(1 + T80)
# Final audited T80 dataset
# ============================================================

LABEL_FILE = os.path.join(
    WORK_DIR,
    "interim",
    "t80_labels.parquet"
)

# Load the FINAL corrected labels (1835 devices)
labels_final = pd.read_parquet(LABEL_FILE)

log_data = labels_final["T80_log1p"].dropna()

# Percentiles used in the manuscript
q25 = log_data.quantile(0.25)
median_val = log_data.median()
q75 = log_data.quantile(0.75)

print("Number of devices:", len(log_data))
print("25th percentile:", q25)
print("Median:", median_val)
print("75th percentile:", q75)

# ------------------------------------------------------------
# Plot
# ------------------------------------------------------------

plt.figure(figsize=(6.5, 4.5))

plt.hist(
    log_data,
    bins=35,
    edgecolor="black",
    linewidth=0.6,
    alpha=0.85
)

plt.axvline(
    q25,
    linestyle="--",
    linewidth=1.5,
    label=f"25th percentile = {q25:.2f}"
)

plt.axvline(
    median_val,
    linestyle="-",
    linewidth=1.8,
    label=f"Median = {median_val:.2f}"
)

plt.axvline(
    q75,
    linestyle="--",
    linewidth=1.5,
    label=f"75th percentile = {q75:.2f}"
)

plt.xlabel("ln(1 + T$_{80}$)")
plt.ylabel("Count")

# No unnecessary title if this will be panel (c) of Figure 1
plt.grid(axis="y", linestyle="--", alpha=0.25)

plt.legend(
    frameon=False,
    fontsize=9
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        ARTIFACT_DIR,
        "Fig1c_T80_log1p_distribution.png"
    ),
    dpi=600,
    bbox_inches="tight"
)

plt.savefig(
    os.path.join(
        ARTIFACT_DIR,
        "Fig1c_T80_log1p_distribution.pdf"
    ),
    bbox_inches="tight"
)

plt.show()
```

```python
if len(t80_cols) > 0:
    target_col = t80_cols[0]

    data = df[target_col].dropna()
    data = data[data > 0]

    lower_limit = data.quantile(0.95)

    fig, ax = plt.subplots(figsize=(6,4))

    ax.hist(data, bins=50, edgecolor='black', alpha=0.8)
    ax.set_title("Distribution of T80 with Highlighted High-Stability Region")
    ax.set_xlabel("T80 (hours)")
    ax.set_ylabel("Count")
    ax.grid(True, linestyle="--", linewidth=0.5, alpha=0.6)

    from mpl_toolkits.axes_grid1.inset_locator import inset_axes

    ax_inset = inset_axes(ax, width="40%", height="40%", loc='upper right')
    ax.axvline(lower_limit, color='red', linestyle='--', linewidth=1.5)

    high_data = data[data >= lower_limit]

    ax_inset.hist(high_data, bins=15, edgecolor='black', alpha=0.8)

    
    ax_inset.tick_params(axis='both')

    plt.tight_layout()
        
    plt.savefig(os.path.join(ARTIFACT_DIR, "t80_distribution_with_inset.png"), dpi=600)
    plt.show()
```

```python
devices_per_doi = df.groupby("Ref_DOI_number").size()

print("Total DOIs:", len(devices_per_doi))
print("Mean devices per DOI:", round(devices_per_doi.mean(), 2))

single_device_ratio = (devices_per_doi == 1).mean()
print(f"Fraction of DOIs with single device: {single_device_ratio:.2f}")

plt.figure(figsize=(6,4))

bins = np.arange(0.5, devices_per_doi.max() + 1.5, 1)

plt.hist(
    devices_per_doi,
    bins=bins,
    edgecolor='black',
    alpha=0.8
)

plt.axvline(
    1,
    color='red',
    linestyle='--',
    linewidth=1.8,
    label='Single-device DOI'
)

ymax = plt.ylim()[1]
plt.text(
    x=1.8,
    y=ymax * 0.85,
    s=f"{single_device_ratio*100:.1f}% of DOIs contain only one device",
    fontsize=10,
    bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray')
)

plt.xlabel("Number of devices per DOI")
plt.ylabel("Number of DOIs")
plt.title("Distribution of Device Counts Across DOI Groups")

plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.legend()
plt.tight_layout()

plt.savefig(os.path.join(ARTIFACT_DIR, "devices_per_doi_distribution.png"), dpi=600)
plt.show()
```
